The code does not change the source workbook. It creates separate dictionaries with pandas DataFrames that will be used for analysis.

```python
# Define a helper function to import data from the Excel file
def split_sheet(data_path, sheet_name, dividers):
    splits = {}
    raw = pd.read_excel(data_path, sheet_name=sheet_name, header=None)

    for start, end in zip(dividers[:-1], dividers[1:]):
        if start < 0 or end > raw.shape[1] or start >= end:
            raise ValueError(f"Invalid block range: {start}:{end}")

        if pd.notna(name := raw.iloc[1, start]):
            block = raw.iloc[3:, start:end].copy()
            block.columns = raw.iloc[2, start:end]
            splits[str(name).strip()] = (block.dropna(how="all").reset_index(drop=True))

    return splits


rockwell_calibration = split_sheet(data_path,"Rockwell Calibration",[0,3,6,9,12])
vickers_data = split_sheet(data_path,"Vickers Hardness",[0,2,4,6])
rockwell_data = split_sheet(data_path,"Rockwell Hardness",[0,2,4,6,8])

print("Loaded pandas DataFrames:")
for what, data in [("Rockwell Calibration", rockwell_calibration),
                   ("Vickers Hardness", vickers_data),
                   ("Rockwell Hardness", rockwell_data)]:
    print(f"{what}: {list(data)}")
    for key, df in data.items():
        print(f"\n{key}:")
        display(df.head(3))
```

Run the code below to generate a plot of the hardness values for the different materials. Update the last line of code to give a suitable caption.

```python
# Plot Rockwell hardness values with standard deviations, separated by scale
fig, axes = plt.subplots(1, 2, figsize=(11, 5), constrained_layout=True)

# Filter the data based on b or c scale
for ax, scale in zip(axes, ["C", "B"]):
    subset = rockwell_statistics[
        rockwell_statistics["Rockwell scale"].str.contains(
        rf"\b{scale}\b",
        regex=True,
        case=False,
        na=False,
        )
    ]

    if subset.empty:
        ax.set_visible(False)

    else:
        plotted_min = (subset["average"] - subset["standard deviation"]).min()
        plotted_max = (subset["average"] + subset["standard deviation"]).max()
        padding = max(5, 0.05 * (plotted_max - plotted_min))
        lower_limit = min(0, plotted_min - padding)
        upper_limit = max(100,plotted_max + padding)

        # Plot all individual measurements for this Rockwell scale
        for i,(material, df) in enumerate(rockwell_data.items()):
            values = pd.to_numeric(df[df["Rockwell Scale"]==scale]["Value"]).dropna()

            ax.scatter(
                [material] * len(values),
                values,
                color="k",
                alpha=0.3,
                s=25,
                label="Raw Data" if i==0 else "_nolegend_",
            )


        ax.errorbar(
            subset.index,
            subset["average"],
            yerr=subset["standard deviation"],
            fmt=" ",
            capsize=5,
            markersize=5,
            label="Mean ± standard deviation",
        )

        ax.set_title(f"Rockwell {scale}")
        ax.set_ylabel(f"Hardness / HR{scale}")
        ax.grid(axis="y", alpha=0.3)
        ax.tick_params(axis="x", rotation=90)
        ax.set_ylim(lower_limit, upper_limit)
        ax.legend()

fig.suptitle("Rockwell Hardness Data")
plt.show()

print("Figure 2. STUDENT: GIVE A SUITABLE CAPTION.")       # STUDENT: provide suitable caption for the plot
```
