### 2.3 Load your data into this Notebook

Run the code cell below. You are **not expected to write or memorize this code**. Read the comments to understand each section and run it without alterations.

```python
# Define a function to read only the measurement runs from each Instron CSV file.
def read_instron_runs(path):
    with path.open(newline="", encoding="utf-8-sig") as handle:
        content = list(csv.reader(handle))

    begin = [
        i for i, line in enumerate(content)
        if len(line) >= 4
        and line[0].strip() in {"1", "2", "3", "4"}
        and line[1].strip() == "Time"
    ]
    end = begin[1:]+[len(content)-1]

    if len(begin) != 4:
        raise ValueError(f"{path.name}: expected four measurement blocks but found {len(begin)}")

    runs = []

    for i in range(len(begin)):
        runs.append(pd.DataFrame(
            np.array([list(map(float,l[1:4])) for l in content[begin[i]+2:end[i]]]),
            columns=["Time/s", "Displacement/mm", "Force/N"]
        ))

    return runs

# Convert the manually entered specimen dimensions into metadata tables
# and pair each of them with the imported measurement runs.

# define a datasets dictionary
datasets = {}


# add the specimen dimensions and the run data to the datasets dictionary
for material, path in FILES.items():
    metadata = pd.DataFrame([
        {
            "Run": specimen["label"],
            "Width": specimen["width"],
            "Thickness": specimen["thickness"],
            "Length": specimen["gauge_length"],
        }
        for specimen in SPECIMEN_DIMENSIONS[material]
    ])

    datasets[material] = (metadata, read_instron_runs(path))

print(
    f"Loaded {len(datasets)} materials and "
    f"{sum(len(metadata) for metadata, runs in datasets.values())} runs."
)
```

```python
# define a function to find the intercept between the 0.2% offset of the linear fit and the data
def find_first_zero_crossing(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    # Exact zero check
    exact_zero = np.where(np.isclose(y, 0.0))[0]
    if exact_zero.size > 0:
        return float(x[exact_zero[0]])

    # Find the first sign change
    for i in range(len(x) - 1):
        y1 = y[i]
        y2 = y[i + 1]

        if (y1 < 0 and y2 > 0) or (y1 > 0 and y2 < 0):
            # Linear interpolation for y = 0 between (x1, y1) and (x2, y2):
            # y = y1 + (y2 - y1) * (x - x1) / (x2 - x1)
            # 0 = y1 + (y2 - y1) * (x - x1) / (x2 - x1)
            denominator = y2 - y1
            if np.isclose(denominator, 0.0):
                return float(x[i])
            x_zero = x[i] - y1 * (x[i + 1] - x[i]) / denominator
            return float(x_zero)

    return None


# determine the yield strain with the function defined above
x = sample["Engineering_Strain"]
y = sample["Engineering_Stress/MPa"]
difference = y - (modulus * (x - offset) + intercept)

yield_strain = find_first_zero_crossing(x,difference)

if yield_strain:
    yield_stress = modulus * (yield_strain - offset) + intercept
    print(f"Intersection strain: {yield_strain:.4f}")
    print(f"Intersection stress: {yield_stress:.0f} MPa")
else:
    yield_strain = np.nan
    yield_stress = np.nan
    print("No intersection found.")
```

```python
def fit_initial_elastic_region(data, lower_stress, upper_stress):
    if lower_stress >= upper_stress:
        raise ValueError("The lower stress limit must be less than the upper limit.")

    stress = data["Engineering_Stress/MPa"]
    maximum_position = stress.to_numpy().argmax()

    mask = (
        stress.between(lower_stress, upper_stress)
        & (np.arange(len(data)) < maximum_position)
    )

    if mask.sum() < 2:
        raise ValueError("The selected stress range contains fewer than two data points.")

    slope, intercept = np.polyfit(
        data.loc[mask, "Engineering_Strain"],
        data.loc[mask, "Engineering_Stress/MPa"],
        1
    )

    return slope, intercept


def find_offset_yield(data, slope, intercept):
    x = data["Engineering_Strain"].to_numpy()
    y = data["Engineering_Stress/MPa"].to_numpy()

    difference = y - (slope * (x - 0.002) + intercept)
    crossings = np.flatnonzero(
        (x[:-1] >= 0.002)
        & (difference[:-1] * difference[1:] <= 0)
    )

    if len(crossings) == 0:
        return np.nan

    i = crossings[0]
    fraction = difference[i] / (difference[i] - difference[i + 1])
    return y[i] + fraction * (y[i + 1] - y[i])

records = []

for mat, (_, runs) in datasets.items():

    lower, upper = FIT_STRESS_LIMITS[material]

    if lower is None:
            lower = stress.min()
    
    if upper is None:
            upper = stress.max()
    

    slope, intercept = fit_initial_elastic_region(runs[0],lower, upper)


    records.append({
        "Material": mat,
        "Fit range / MPa": f"{FIT_STRESS_LIMITS[mat][0]}–{FIT_STRESS_LIMITS[mat][1]}",
        "Yield strength / MPa": find_offset_yield(runs[0], slope, intercept),
        "Young's modulus / GPa": slope / 1000,
    })

property_table = pd.DataFrame(records)
display(property_table.round(2))


```
